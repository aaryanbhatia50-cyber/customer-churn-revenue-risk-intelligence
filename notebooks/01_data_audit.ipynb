```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

```python
customer_df = pd.read_csv("../data/raw/customer_data.csv")

transactions_df = pd.read_csv(
    "../data/raw/transactions_data.csv",
    parse_dates=["date"]
)
```

```python
print("Customer Dataset:", customer_df.shape)
print("Transaction Dataset:", transactions_df.shape)

display(customer_df.head())
display(transactions_df.head())
```

```python
customer_df.info()
```

```python
transactions_df.info()
```

```python
missing_values = pd.DataFrame({
    "Customer Missing": customer_df.isna().sum(),
    "Transaction Missing": transactions_df.isna().sum()
})

missing_values = missing_values[
    (missing_values["Customer Missing"] > 0) |
    (missing_values["Transaction Missing"] > 0)
]

if missing_values.empty:
    print("No missing values found.")
else:
    display(missing_values)
```

```python
print("Duplicate customer IDs:", customer_df["customer_id"].duplicated().sum())
print("Duplicate transaction rows:", transactions_df.duplicated().sum())

print("Invalid transaction dates:", transactions_df["date"].isna().sum())
print("Zero-value transactions:", (transactions_df["amount"] == 0).sum())
print("Negative-value transactions:", (transactions_df["amount"] < 0).sum())

print(
    "Transactions without matching customers:",
    (~transactions_df["customer_id"].isin(customer_df["customer_id"])).sum()
)
```

```python
print("Transaction Date Range:")
print(transactions_df["date"].min(), "to", transactions_df["date"].max())

display(transactions_df["amount"].describe().to_frame("Amount"))
```

```python
display(
    transactions_df["amount"]
    .quantile([0.90, 0.95, 0.99, 0.995, 0.999])
    .to_frame("Transaction Amount")
)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(transactions_df["amount"], bins=50)
plt.title("Distribution of Transaction Amounts")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
display(transactions_df["type"].value_counts().to_frame("Transaction Count"))
```

```python
categorical_columns = customer_df.select_dtypes(
    include=["object"]
).columns

display(customer_df[categorical_columns].describe().T)
```

```python
product_columns = [
    "savings_account",
    "credit_card",
    "personal_loan",
    "investment_account",
    "insurance_product"
]

display(
    customer_df[product_columns]
    .sum()
    .sort_values(ascending=False)
    .to_frame("Customer Count")
)
```
