```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

```python
customer_df = pd.read_csv("../data/raw/customer_data.csv")

transactions_df = pd.read_csv(
    "../data/raw/transactions_data.csv",
    parse_dates=["date"]
)

print("Customer Dataset:", customer_df.shape)
print("Transaction Dataset:", transactions_df.shape)
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(customer_df["age"], bins=30, kde=True)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=customer_df, x="gender")

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=customer_df,
    x="income_bracket",
    order=customer_df["income_bracket"].value_counts().index
)

plt.title("Customer Distribution by Income Bracket")
plt.xlabel("Income Bracket")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
product_columns = [
    "savings_account",
    "credit_card",
    "personal_loan",
    "investment_account",
    "insurance_product"
]

product_counts = (
    customer_df[product_columns]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(9, 5))

sns.barplot(
    x=product_counts.index,
    y=product_counts.values
)

plt.title("Customer Adoption by Product")
plt.xlabel("Product")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    customer_df["app_logins_frequency"],
    bins=30,
    kde=True
)

plt.title("Distribution of App Login Frequency")
plt.xlabel("App Login Frequency")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    customer_df["feature_usage_diversity"],
    bins=np.arange(-0.5, 11.5, 1)
)

plt.title("Distribution of Feature Usage Diversity")
plt.xlabel("Number of Features Used")
plt.ylabel("Number of Customers")
plt.xticks(range(11))
plt.tight_layout()
plt.show()
```

```python
transaction_volume_by_type = (
    transactions_df
    .groupby("type")["amount"]
    .agg(
        transaction_count="count",
        total_volume="sum",
        average_amount="mean"
    )
    .sort_values("total_volume", ascending=False)
)

display(transaction_volume_by_type.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=transactions_df,
    x="type",
    order=transactions_df["type"].value_counts().index
)

plt.title("Transaction Count by Type")
plt.xlabel("Transaction Type")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=transaction_volume_by_type.reset_index(),
    x="type",
    y="total_volume"
)

plt.title("Total Transaction Volume by Type")
plt.xlabel("Transaction Type")
plt.ylabel("Total Transaction Volume")
plt.tight_layout()
plt.show()
```

```python
monthly_transactions = (
    transactions_df
    .set_index("date")
    .resample("ME")
    .agg(
        transaction_count=("amount", "count"),
        total_volume=("amount", "sum")
    )
    .reset_index()
)
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=monthly_transactions,
    x="date",
    y="transaction_count",
    marker="o"
)

plt.title("Monthly Transaction Count")
plt.xlabel("Month")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=monthly_transactions,
    x="date",
    y="total_volume",
    marker="o"
)

plt.title("Monthly Transaction Volume")
plt.xlabel("Month")
plt.ylabel("Total Transaction Volume")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
